```
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload image
uploaded = files.upload()
file = list(uploaded.keys())[0]

# Step 2: Read image in grayscale
img = cv2.imread(file, 0)

# Step 3: Apply Histogram Equalization
eq = cv2.equalizeHist(img)

# Step 4: Display Original & Equalized
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Equalized Image")
plt.imshow(eq, cmap='gray')
plt.axis('off')

plt.show()

# Step 5: Histogram Graph (Comparison)
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("Original Histogram")
plt.hist(img.ravel(), bins=256)

plt.subplot(1,2,2)
plt.title("Equalized Histogram")
plt.hist(eq.ravel(), bins=256)

plt.show()
```

```
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload image
uploaded = files.upload()
file = list(uploaded.keys())[0]

# Step 2: Read image (color)
img = cv2.imread(file)

# Step 3: Convert to YCrCb
ycrcb = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)

# Step 4: Apply CLAHE (better contrast)
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
ycrcb[:,:,0] = clahe.apply(ycrcb[:,:,0])

# Step 5: Convert back to RGB
result = cv2.cvtColor(ycrcb, cv2.COLOR_YCrCb2RGB)

# Step 6: Display images
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Enhanced Image (CLAHE)")
plt.imshow(result)
plt.axis('off')

plt.show()

# Step 7: Histogram comparison
gray_orig = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray_res = cv2.cvtColor(result, cv2.COLOR_RGB2GRAY)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("Original Histogram")
plt.hist(gray_orig.ravel(), bins=256)

plt.subplot(1,2,2)
plt.title("Enhanced Histogram")
plt.hist(gray_res.ravel(), bins=256)

plt.show()
```
